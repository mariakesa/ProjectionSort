```python
import numpy as np
import matplotlib.pyplot as plt
n_samples = 2000
d_latent  = 10

X = np.random.randn(n_samples, d_latent)
true_w = np.random.randn(d_latent)
y = (X @ true_w > 0).astype(int)
plt.plot(y[:100])
```

```python
n_neurons = 400
W = np.random.randn(d_latent, n_neurons)

R = np.tanh(X @ W)
```

```python
from sklearn.decomposition import PCA

pca = PCA()
Z = pca.fit_transform(R)
```

```python
from sklearn.linear_model import LogisticRegression

clf = LogisticRegression(max_iter=500)
clf.fit(Z, y)

beta = clf.coef_[0]
```

```python
plt.stem(np.abs(beta[:50]))
```

```python
import numpy as np
from sklearn.decomposition import PCA

# nonlinear mixed responses
x = np.random.randn(1000, 2)
r1 = x[:,0]
r2 = x[:,1]
r3 = x[:,0] * x[:,1]   # nonlinear mixed-selective neuron

R = np.vstack([r1, r2, r3]).T
p = PCA().fit(R)

print(p.explained_variance_ratio_)
```
