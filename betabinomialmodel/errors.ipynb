```python
import os
import numpy as np
import matplotlib.pyplot as plt

IMG_DIR = "/home/maria/ProjectionSort/data/images/images"
ERR_PATH = "/home/maria/ProjectionSort/data/error_matrix.npy"
OUT_DIR = "/home/maria/ProjectionSort/data/error_viz"

os.makedirs(OUT_DIR, exist_ok=True)

names = ["BetaNB", "Logistic", "SVM"]

E = np.load(ERR_PATH)
n = E.shape[0]

# ---------------------------------------------------
# Helper to draw a grid of images
# ---------------------------------------------------
def plot_images(idxs, title, fname, ncols=6):
    if len(idxs) == 0:
        print(f"No images for {title}")
        return

    nrows = int(np.ceil(len(idxs) / ncols))
    fig, axes = plt.subplots(nrows, ncols, figsize=(ncols*2, nrows*2))
    axes = np.atleast_2d(axes)

    for ax in axes.flat:
        ax.axis("off")

    for k, i in enumerate(idxs):
        r, c = divmod(k, ncols)
        img_path = os.path.join(IMG_DIR, f"scene_{i:03d}.png")
        img = plt.imread(img_path)
        axes[r, c].imshow(img)
        axes[r, c].set_title(f"{i}", fontsize=7)

    plt.suptitle(title)
    plt.tight_layout()
    plt.savefig(os.path.join(OUT_DIR, fname), dpi=200)
    plt.close()
    print(f"Saved {fname}")

# ---------------------------------------------------
# Image groups
# ---------------------------------------------------
all_wrong = np.where(E.sum(axis=1) == 3)[0]

beta_only = np.where((E[:,0] == 1) & (E[:,1] == 0) & (E[:,2] == 0))[0]
log_only  = np.where((E[:,0] == 0) & (E[:,1] == 1) & (E[:,2] == 0))[0]
svm_only  = np.where((E[:,0] == 0) & (E[:,1] == 0) & (E[:,2] == 1))[0]

beta_log_wrong = np.where((E[:,0] == 1) & (E[:,1] == 1) & (E[:,2] == 0))[0]
beta_svm_wrong = np.where((E[:,0] == 1) & (E[:,1] == 0) & (E[:,2] == 1))[0]
log_svm_wrong  = np.where((E[:,0] == 0) & (E[:,1] == 1) & (E[:,2] == 1))[0]

print("Counts:")
print("All wrong:", len(all_wrong))
print("Beta only:", len(beta_only))
print("Log only :", len(log_only))
print("SVM only :", len(svm_only))

# ---------------------------------------------------
# Plot
# ---------------------------------------------------
plot_images(all_wrong, "All models wrong (biological blind spots)", "all_wrong.png")
plot_images(beta_only, "Only BetaNB wrong", "beta_only.png")
plot_images(log_only, "Only Logistic wrong", "log_only.png")
plot_images(svm_only, "Only SVM wrong", "svm_only.png")

plot_images(beta_log_wrong, "BetaNB + Logistic wrong", "beta_log_wrong.png")
plot_images(beta_svm_wrong, "BetaNB + SVM wrong", "beta_svm_wrong.png")
plot_images(log_svm_wrong,  "Logistic + SVM wrong", "log_svm_wrong.png")
```
